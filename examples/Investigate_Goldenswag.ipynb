```python
import sys
sys.path.append('/Users/davebrunner/Documents/repositories/SelfClean/')
import time
from pathlib import Path
from selfclean import SelfClean
from selfclean.cleaner.issue_manager import IssueTypes
from selfclean.core.src.pkg.helper import EmbeddingPoolingType
```

# Running and evaluating contaminated GoldenSwag files
This notebook shows how for every issue type the synthetic contamination can be used to evaluate and generate performance metrics

# Near Duplicates

```python
start_time = time.time()
selfclean = SelfClean(
    plot_top_N=10,
    auto_cleaning=True,
    output_path="../examples/output/Goldenswag",
)
results, prediction = selfclean.run_on_text_dataset(
    dataset_path=Path('../experiment/datasets/goldenswag/golden_swag_train_synthetic_NEAR_DUPLICATES.json'),
    contamination_log_path=Path('../experiment/datasets/goldenswag/golden_swag_train_synthetic_NEAR_DUPLICATES_logs.json'),
    pretraining_type="electra",
    epochs=0,
    batch_size=32,
    cache_dir=None,
    dataset_name="hellaswag",
    base_model="golden_swag_train_electra_NEAR_DUPLICATE_35",
    wandb_logging=False,
    issues_to_detect=[
        IssueTypes.NEAR_DUPLICATES,
    ],
    pooling_type=EmbeddingPoolingType.MEAN,
)
print(f"\nFinished in {(time.time() - start_time) / 60:.2f} minutes")
```
